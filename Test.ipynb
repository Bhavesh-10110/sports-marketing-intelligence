```python
import pandas as pd

customer_df = pd.read_csv(r"data/raw/Customer_Dataset.csv")
engagement_df = pd.read_csv(r"data/raw/ipl_engagement_dataset.csv")
social_df = pd.read_csv(r"data/raw/social_media_dataset.csv")
purchase_df = pd.read_csv(r"data/raw/purchase_dataset.csv")
```

```python
# Merge datasets using customer_id
master_df = customer_df.merge(
    engagement_df,
    on="customer_id"
).merge(
    social_df,
    on="customer_id"
).merge(
    purchase_df,
    on="customer_id"
)

# Save merged dataset
master_df.to_csv(
    "data/processed/master_dataset.csv",
    index=False
)

print("Datasets merged successfully!")
master_df.head()
```

```python
df = pd.read_csv(
    "data/processed/processed_dataset.csv"
)
```

```python
cluster_features = [

    "age",
    "monthly_income",
    "matches_watched_per_season",
    "sports_content_watch_hours_per_week",
    "sports_equipment_spending_per_month"
]
X = df[cluster_features]
```

```python

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

df["cluster"] = kmeans.fit_predict(X)
```

```python
score = silhouette_score(
    X,
    df["cluster"]
)

print(f"\nSilhouette Score: {score:.3f}")
```
